```python
# TODO 1 — define representative inputs and expected outputs
def add(a:float, b:float)->float:
    """return the sum of two numbers."""
    return a + b

def substract(a: float, b: float) -> float:
    """Return the difference of two numbers."""
    return a - b

def multiply(a: float, b: float) -> float:
    """Return the product of two numbers."""
    return a * b

def divide(a: float, b: float) -> float:
    """Return the quotient of two numbers."""
    if b == 0:
        raise ValueError("Cannot divide by zero.")
    return a / b
```

```python
# TODO 2 — implement the smallest end-to-end path

a = 150
b = 50
c = add(a, b)

assert c == 200
assert substract(a, b) == 100
assert multiply(a, b) == 7500
assert divide(a, b) == 3
assert add(a, c) == 350
assert substract(c, b) == 150

print("All tests passed! You can Start!")
print("get ready in 5!")
print("4")
print("3")
print("2")
print("1")

#Hasta aqui lo he hecho yo, lo que sigue lo he gestionado con IA pero basandome en las notas de clase. Pero entiendo los conceptos basicos del loop y el brake asi como la definicion de funciones y el uso de assert para testear.

def main():
    while True:
        print("")
        print("===== CALCULATOR =====")
        print("1. Add")
        print("2. Substract")
        print("3. Multiply")
        print("4. Divide")
        print("5. Exit")
        print("=======================")

        option = input("Choose an option (1-5): ")

        if option == "5":
            print("Bye Bye!")
            break

        num1 = float(input("Introduce the first number: "))
        num2 = float(input("Introduce the second number: "))

        if option == "1":
            print("Result:", add(num1, num2))
        elif option == "2":
            print("Result:", substract(num1, num2))
        elif option == "3":
            print("Result:", multiply(num1, num2))
        elif option == "4":
            print("Result:", divide(num1, num2))
        else:
            print("Invalid option. Please try again.")


main()

```
